```python
# Creating a list
fruits = ["apple", "banana", "cherry"]

# Accessing elements
print(fruits[0])     # apple
print(fruits[-1])    # cherry

# Modifying elements
fruits[1] = "mango"
print(fruits)        # ['apple', 'mango', 'cherry']

# Adding elements
fruits.append("orange")
print(fruits)        # ['apple', 'mango', 'cherry', 'orange']

# Removing elements
fruits.remove("apple")
print(fruits)        # ['mango', 'cherry', 'orange']
```

```python
# Built-in Functions: len(), max(), min(), sum(), sorted(), any(), all(), tuple()
t = (10, 20, 30, 40)
len(t)       # Returns length of tuple
max(t)       # Returns maximum value
min(t)       # Returns minimum value
sum(t)       # Returns sum of elements (if numbers)
sorted(t)    # Returns a sorted list from tuple
any(t)       # Returns True if any element is True
all(t)       # Returns True if all elements are True
# tuple(seq)   # Converts an iterable into a tuple

print(len(t))       # 4
print(max(t))       # 40
print(min(t))       # 10
print(sum(t))       # 100
print(sorted(t))    # [10, 20, 30, 40] (list, not tuple)
print(any(t))       # True (since elements are non-zero)
print(all(t))       # True (all non-zero)
print(tuple("abc")) # ('a', 'b', 'c')
```

<h3 style="color:brown">(a) Modulus/Modulo(%) Operator</h3>

## It gives the remainder when one number is divided by another

```python
# module operator which return -------remiander value
u = int(input("Enter your number 1: "))
n = int(input("Enter you number 2: "))  
c = u%n
print("Remiander of c is:",c)
```

```python
print(10 % 3)   # 1 (10 ÷ 3 = 3 remainder 1)
print(15 % 4)   # 3 (15 ÷ 4 = 3 remainder 3)
print(8 % 2)    # 0 (8 is exactly divisible by 2)
```

```python
# 🔹 Works with Negative Numbers too
print(-10 % 3)   # 2
print(10 % -3)   # -2
```

```python
# Extra:- % is also used in old-style string formatting.
name = "Ruchi"
age = 15
print("My name is %s and I am %d years old." % (name, age))
# Output: My name is Ruchi and I am 15 years old.
```

<h2 style="color:brown">(b) Floor Devision (//) </h2>

```python
print(10 // 3)   # 3  (10 ÷ 3 = 3.33 → floor = 3)
print(20 // 7)   # 2  (20 ÷ 7 = 2.85 → floor = 2)
print(8 // 2)    # 4  (exact division)
```

```python
print(-10 // 3)   # -4  (since -3.33 floored = -4, not -3)
print(10 // -3)   # -4
print(-10 // -3)  # 3
# 👉 Important: Unlike normal integer division (which just truncates), floor division always goes down (towards negative infinity)
```

```python
# With floats 
print(7.5 // 2)   # 3.0  (float result because operand is float)
print(7 // 2.5)   # 2.0
```

```python
# Floor Devision which return ----- value
u = int(input("Enter your number1: "))
n = int(input("enter you number 2: ")) 
v = u//n
print("the floor devision is :",v)
print(type(v))
```

<h2 style="color:brown">(c) Comperision Operator</h2>

### Comparison operators are used to compare two values.
They always return a Boolean value → True or False.

```python
a = 34
b = 80
print(a>b)    # False
print(a != b)   # True
print(a > b)    # False
print(a < b)    # True
print(a >= 10)  # True
print(b <= 15)  # False
```

### 🔹 With Strings

Python compares strings lexicographically (alphabetical order, based on Unicode values).

```python
print("apple" == "apple")   # True
print("apple" != "banana")  # True
print("apple" < "banana")   # True ('a' < 'b')
```

### 🔹 Difference Between /, //, and %

/ → Normal division (returns float).

// → Floor division (integer/floor value).

% → Modulus (remainder).


```python
print(10 / 3)   # 3.333...
print(10 // 3)  # 3
print(10 % 3)   # 1
```
